## Объявление и тело цикла

При объявлении цикла создается служебная переменная, которая для каждого из повторений тела цикла будет принимать новое значение. Например, такую переменную можно назвать просто "x".

```python
for x in bremen_musicians: # объявление цикла
    print(x) # тело цикла
```

Использование цикла for в данном случае позволяет форматировать строковые значения внутри списка при выводе их на экран.

```python
for i, item in enumerate(bremen_musicians):
    print(f'Музыкант {i + 1}: {item}')
```

Кроме работы со списками, циклы используются и для повторения действий множественное количество раз. При помощи цикла for и функции range(), например, можно несколько раз отобразить на экране заданное сообщение.

Функция range() принимает два аргумента – верхнюю и нижнюю границу, при количество повторений будет равняться разнице между этими двумя значениями. Также можно обозначить и третий аргумент, который установит шаг между пограничными значениями.

```python
for i in range(1, 11):
  print('test')
```

```python
for i in range(1, 11, 2):
  print(i)
```

## Множественные ветвления

Базовую конструкцию ветвления if-else можно усложнить несколькими способами, чтобы создать ветвление с несколькими условиями или большим количеством альтернатив.



```python
var = '100500'
if type(var) is int:
  print('Это число')
  res = var * 2
if type(var) is str:
    print('Это строка')
    res = var + ' - строка!'
if type(var) is bool:
    print('Это логический тип')
    var = not var
print(res)
```

Для остановки выполнения ветвления там, где удовлетворяется условие, можно использовать оператор elif (else + if). Логика ветвления при этом не изменяется за исключением того, что объявленные через elif части будут выполняться, только если предыдущие ветки не были отработаны.

```python
if type(var) is int:
  print('Это число')
  res = var * 2
elif type(var) is str:
    print('Это строка')
    res = var + ' - строка!'
elif type(var) is bool:
    print('Это логический тип')
    res = not var
print(res)
```

Программу с ветвлением можно сделать более предсказуемой, добавив после if и elif еще и часть с else – то есть обозначить вариант выполнения, если ни одно из ожидаемых нами условий не было удовлетворено.

В следующем примере вводная переменная var задана списочным типом, а значит не удовлетворяет ни одному из вариантов внутри ветвления. И переменной res будет присвоено значение None, что предусмотрено в части else ветвления.

```python
var = [100]
if type(var) is int:
  print('Это число')
  res = var * 2
elif type(var) is str:
    print('Это строка')
    res = var + ' - строка!'
elif type(var) is bool:
    print('Это логический тип')
    res = not var
else:
  print('Какой-то неизвестный тип...')
  res = None
print(res)
```

Если же переменная var будет задана как целое число, то выполнена будет только первая часть ветвления, удовлетворяющая условию type(var) is int. Переменная res в этом случае будет равна вводной переменной, умноженной на 2.

```python
var = 200
if type(var) is int:
  print('Это число')
  res = var * 2
elif type(var) is str:
    print('Это строка')
    res = var + ' - строка!'
elif type(var) is bool:
    print('Это логический тип')
    res = not var
else:
  print('Какой-то неизвестный тип...')
  res = None
print(res)
```

Пример

```python
def div(a, b):
  res = None
  try:
    res = a / b
  except TypeError:
    print('Похоже, вы используете несовместимые типы...')
  except:
    print('Ошибка')
  return res
```

```python
div(100, 10)
```

```python
div(100, 0)
```

```python
div(100, '10')
```

Можно попробовать перехватить еще больше типов ошибок и усложнить логику функции далее. Новый вид функции div() сможет отреагировать не только на TypeError, но и на ZeroDivisionError.

Ключевое слово else поможет сообщить об успешном выполнении деления, если не возникло ошибок из блоков except. Часть программы после finally будет выполняться вне зависимости от того, по какому из вариантов ветвления пройдет программа.

```python
def div(a, b):
  res = None
  try:
    res = a / b
  except TypeError:
    print('Похоже, вы используете несовместимые типы...')
  except ZeroDivisionError:
    print('На ноль делить нельзя!')
  else:
    print('Все хорошо!')
  finally:
    print('Выполнение программы завершено')
  return res
```

```python
div(100, 10)
```

```python
div(100, 0)
```

# Практическая работа 1

```python
class Chocolate:
  def __init__(self,price:float,additive:str):
    self.price=price
    self.additive=additive

    if price <=0:
      raise ValueError("Цена батончика должна быть больше нуля!")

  def show_my_chocolate(self):
    if self.additive:
      print(f"Шоколадный батончик, добавка: {self.additive}")
    else:
      print("Обычный шоколадный батончик")

  def __str__(self): # Преобразование к строке str(obj)
    return f"Шоколадный батончик, {self.price} рублей"

```

```python
choc=Chocolate(1,"")
choc.show_my_chocolate()
print(choc)
```

```python
choc1=Chocolate(0,"миндаль")
choc1.show_my_chocolate()
print(choc1)
```

```python
choc2=Chocolate(200,"миндаль")
choc2.show_my_chocolate()
print(choc2)
```
